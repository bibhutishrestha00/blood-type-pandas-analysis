```python
df['A-'] = df['A-'].fillna(df['A-'].mean())
```

```python
df['B-'] = df['B-'].fillna(df['B-'].mean())
```

```python
df['AB-'] = df['AB-'].fillna(df['AB-'].mean())
```

```python
df['O-'] = df['O-'].fillna(df['O-'].mean())
```

```python
df.isnull().sum()
```

```python
df['Continent'].mode()
```

```python
df['Continent'] = df['Continent'].fillna(df['Continent'].mode()[0])
```

```python
df.isnull().sum()
```

```python
df['Total_Positive'] = df['O+'] + df['A+'] + df['B+'] + df['AB+']
```

```python
df['Total_Negative'] = df['O-'] + df['A-'] + df['B-'] + df['AB-']
```

```python
df['Positive_Minus_Negative'] = df['Total_Positive'] - df['Total_Negative']
```

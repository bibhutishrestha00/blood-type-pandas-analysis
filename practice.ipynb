Write a Python program using Pandas to create a table that contains the following columns and data:

Employee Name: John, Kate, Leo

Department: IT, HR, Sales

Salary: 50000, 55000, 60000

```python
import pandas as pd
data = {
    'Name': ['John', 'Kate', 'Leo'],
    'Department': ['IT', 'HR', 'Sales'],
    'Salary': [50000, 55000, 60000]
}
df = pd.DataFrame(data)
print(df)
```

Create a Series, DataFrame from both using dictionary and using list of lists

```python
import pandas as pd
data = {'Class': ['A', 'B', 'C', 'D'],
        'Math': [85, 90, 78, 92],
        'Science': [88, 95, 80, 89],
        'English': [82, 87, 75, 91]}
df = pd.Series(data)
print(df)
```

```python
import pandas as pd
data = [[10, 20], ['abc', 60], ['hello', 100]]
df = pd.Series(data)
print(df)
```

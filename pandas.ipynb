Create a new column called Total_Positive.

It should contain:

O+ + A+ + B+ + AB+

```python
df['Total_Positive'] = df['O+'] + df['A+'] + df['B+'] + df['AB+']
```

Create a new column called Total_Negative containing:

O- + A- + B- + AB-

```python
df['Total_Negative'] = df['O-'] + df['A-'] + df['B-'] + df['AB-']
```

Now compare positive vs negative blood types.

Create a new column called Positive_Minus_Negative:

```python
df['Positive_Minus_Negative'] = df['Total_Positive'] - df['Total_Negative']
```

Create a pivot table showing the average blood-type percentage for each continent.

```python
pivot = df.pivot_table(
    index = 'Continent',
    values = ['O+', 'A+', 'B+', 'AB+', 'O-', 'A-', 'B-', 'AB-'],
    aggfunc = 'mean'
)
pivot
```

Fill all missing values in the A+ column with its average value.

```python
df['A+'] = df['A+'].fillna(df['A+'].mean())
```
